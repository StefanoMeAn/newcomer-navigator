## Introduction

This notebook defines the first version of Newcomer Navigator, an English-first guide to official procedures for people settling in Italy. It records the project boundaries and creates a traceable register of candidate sources before their content is used.

## Learning objectives

- Define what the first version of the assistant will and will not do.
- Record official sources with enough metadata to review them.
- Use pandas to create, validate, and save a source register.

## Initial scope

The first version focuses on public information for Padova and three procedures:

- residence registration;
- requesting a Carta d'Identità Elettronica (CIE);
- understanding the official ISEE process.

The assistant will explain published procedures and link to their sources. It will not submit applications, determine legal eligibility, or store identity documents or personal circumstances.

## Source and translation rules

The original Italian text remains the authoritative source. English explanations are unofficial translations. Answers must link to the original page, and the assistant must say when its sources do not support an answer.

All sources start with the status `needs_review`. This means they are candidates for inspection, not approved content.

## First source candidates

The initial candidates are the Comune di Padova’s pages about residence registration and CIE, and INPS’s ISEE service page. We’ll verify their scope and currentness before using their content.

```python
from pathlib import Path
import pandas as pd

project_root = Path.cwd()
if project_root.name == "notebooks":
    project_root = project_root.parent

sources = pd.DataFrame(
    [
        {
            "source_id": "PADOVA_RESIDENCE",
            "title_original": "Iscrizione anagrafica per cittadini di Stato terzo",
            "title_english": "Residence registration for non-EU citizens",
            "authority": "Comune di Padova",
            "jurisdiction": "Comune di Padova",
            "procedure": "residence registration",
            "source_language": "Italian",
            "url": "https://www.comune.padova.it/servizi/anagrafe-e-stato-civile/cittadine-e-cittadini-di-stato-terzo-non-appartenenti-allunione",
            "date_checked": "2026-10-15",
            "status": "needs_review",
        },
        {
            "source_id": "PADOVA_CIE",
            "title_original": "Carta d'identità elettronica - CIE 3.0",
            "title_english": "Electronic Identity Card (CIE)",
            "authority": "Comune di Padova",
            "jurisdiction": "Comune di Padova",
            "procedure": "CIE",
            "source_language": "Italian",
            "url": "https://www.comune.padova.it/carta-didentita-elettronica-cie-30",
            "date_checked": "2026-10-15",
            "status": "needs_review",
        },
        {
            "source_id": "INPS_ISEE",
            "title_original": "Portale unico ISEE",
            "title_english": "ISEE portal",
            "authority": "INPS",
            "jurisdiction": "Italy",
            "procedure": "ISEE",
            "source_language": "Italian",
            "url": "https://www.inps.it/it/it/dettaglio-scheda.it.schede-servizio-strumento.schede-servizi.Portale-unico-ISEE.html",
            "date_checked": "2026-10-15",
            "status": "needs_review",
        },
    ]
)
```

```python
required_columns = {
    "source_id",
    "title_original",
    "title_english",
    "authority",
    "jurisdiction",
    "procedure",
    "source_language",
    "url",
    "date_checked",
    "status",
}

assert required_columns.issubset(sources.columns)
assert sources["source_id"].is_unique
assert sources["url"].str.startswith("https://").all()
assert sources["status"].eq("needs_review").all()

sources_path = project_root / "data" / "sources.csv"
sources.to_csv(sources_path, index=False)

print(f"Saved {len(sources)} candidate sources to {sources_path}")
```
